```python
import pandas as pd
import numpy as np

df = pd.read_csv("../test_data_export.csv")
df = df.sort_index(axis=1)
df
```

```python
vuf = df['vuf']
# vuf_approx = df['vuf_approx']
faulted = df['faulted']
```

```python
# vuf accuracy
true_values_vuf = df[((vuf >= 0.01) & (faulted == 1)) | ((vuf < 0.01) & (faulted == 0))]
print("accuracy is: ", true_values_vuf.shape[0] / df.shape[0])
true_values_vuf
```
